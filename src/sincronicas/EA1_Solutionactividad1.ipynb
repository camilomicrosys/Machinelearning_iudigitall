```python
# Ejemplo para convertir las columnas que sean arreglos a texto:
import numpy as np

for col in df.columns:
  if df[col].apply(lambda x: isinstance(x, (np.ndarray, list))).any():
    df[col] = df[col].astype(str)
```

```python
# Convierte cualquier columna que tenga diccionarios o listas a formato de texto (string)
for col in df.columns:
  if df[col].apply(lambda x: isinstance(x, (dict, list))).any():
    df[col] = df[col].astype(str)
```
